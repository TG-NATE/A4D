# Structuring Data

### Matching Data From Multiple Sources
when matching data from mutiple sources you must spend time figuring out of the data is compatible. The following steps helps you work through this process.

1. Determine if both sets has all the required data
2. Check if both sets are using same data types for properties
3. Making sure the data elements in both sets share the same meaning
4. Verify the data attributes

### Structural Remedation

#### Data Duplication
Data in sets can become duplicated from combing sets or shaping data.

The Panda package can be use to remove duplicated data aka noise

```python
import pandas as pd

df = pd.DataFrame({ 'A': [0, 0, 0, 0, 0, 1, 0],
                    'B': [0, 2, 3, 5, 0, 2, 0],
                    'C': [0, 3, 4, 1, 0, 2, 0]})

print(df, "\n")

df = df.drop_duplicates()

print(df)
```

#### Missing Values
Missing values can skew data and cause errors so they need to be handled

This can be handled in 2 ways

+ Setting missing data to a std value
    + Works, but could also skew data

+ Setting missing data to a mean of of all values
    + Typically makes missing data not count

```python
df = pd.DataFrame({
    'A' : [0, 0, 1, None],
    'B' : [1, 2, 3, 4],
    'C' : [np.NAN, 3, 4, 2],
}, dtype = int)
```
