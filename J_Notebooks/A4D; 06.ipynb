### Structural Remedation

#### Data Duplication
Data in sets can become duplicated from combing sets or shaping data.

The Panda package can be use to remove duplicated data aka noise

```python
import pandas as pd

df = pd.DataFrame({ 'A': [0, 0, 0, 0, 0, 1, 0],
                    'B': [0, 2, 3, 5, 0, 2, 0],
                    'C': [0, 3, 4, 1, 0, 2, 0]})

print(df, "\n")

df = df.drop_duplicates()

print(df)
```

#### Missing Values
Missing values can skew data and cause errors so they need to be handled

This can be handled in 2 ways

+ Setting missing data to a std value
    + Works, but could also skew data

+ Setting missing data to a mean of of all values
    + Typically makes missing data not count

```python
df = pd.DataFrame({
    'A' : [0, 0, 1, None],
    'B' : [1, 2, 3, 4],
    'C' : [np.nan, 3, 4, 2],
})


print(df, '\n')

values = pd.Series(df.mean(), dtype=int)
print(values, "\n")

df = df.fillna(values).astype(int)
print(df)
```

### Stacks

LIFO data storage structure

2 main processing functions

1. Push - Pushes data to the stack
2. Pop - Pops data from top of stack

```python
MyStack = []

stackSize = 3

def displayStack():
    print("Items currently in stack")
    for item in MyStack:
        print(item)
        
def push(item):
    if len(MyStack) < stackSize:
        MyStack.append(item)
    else:
        print("Stack is full")
        
def pop():
    if len(MyStack) > 0:
        print("Popping: ", MyStack.pop())
    else:
        print("Stack is empty")
        
displayStack()

push(1)
push(2)
push(3)
displayStack()
push(4)



pop()
pop()
pop()
displayStack()
pop()
```

### queues

FIFO data storage structure

2 main processing functions

1. Put - Puts data to the Queue
2. Get - Gets data from front of the queue
3. Empty - Returns T/F based on if the queue is empty
4. Full - Returns T/F based on if the queue is full

```python
import queue

MyQueue = queue.Queue(3)

print("Queue empty: ", MyQueue.empty())

MyQueue.put(1)
MyQueue.put(2)
MyQueue.put(3)

print("Queue full: ", MyQueue.full())

print("popping: ", MyQueue.get())
print("Queue full: ", MyQueue.full())

print("popping: ", MyQueue.get())
print("popping: ", MyQueue.get())
print("Queue full: ", MyQueue.empty())

```

### Dictionaries - Data storage method that works based on key value pairs

Rules
1. The key must be unique
2. The key must be immutable

```python
Colors = {"Sam" : "Blue", "Amy" : "Red", "Sarah" : "Yellow"}

print(Colors["Sarah"])
print(Colors.keys())

for item in Colors.keys():
    print("{0} likes the color {1}.".format(item, Colors[item]))
    
Colors["Sarah"] = "Purple"
Colors.update({"Harry" : "Orange"})
del Colors["Sam"]

print(Colors)
```

```python
class binaryTree:
    def __init__(self, data, left = None, right = None):
        self.data = data
        self.left = left
        self.right = right
        
    ##Builtiin Python Function to when print is called on node data it print the val not the memeory space.
    def __str__(self):
        return str(self.data)
    
    
MyTree = binaryTree("Root")
BranchA = binaryTree("Branch A")
BranchB = binaryTree("Branch B")
MyTree.left = BranchA
MyTree.right = BranchB

LeafC = binaryTree("Leaf A")
LeafD = binaryTree("Leaf D")
LeafE = binaryTree("Leaf E")
LeafF = binaryTree("Leaf F")

BranchA.left = LeafC
BranchA.right = LeafD
BranchB.left = LeafE
BranchB.right = LeafF

def traverseTree(tree):
    if tree.left != None:
        traverseTree(tree.left)
    if tree.right != None:
        traverseTree(tree.right)
    print(tree.data)
    

traverseTree(MyTree)
```
