#### Sorting Data with Quick Sort and Merge Sort

##### Selection Sort
Brute force since its checking every element

Worst-case Speed: O(n<sup>2</sup>)

Best use case: Sorting Linked List

Preforms fewer exchanges than insertion sort

sorted in place

moves smallest item to front of list or moves the biggest item to the back of the list


```python
data = [9, 5, 7, 4, 2, 8, 1, 10, 6, 3]

for scanIndex in range(0, len(data)):
    minIndex = scanIndex
    
    for compIndex in range (scanIndex + 1, len(data)):
        if data[compIndex] < data[minIndex]:
            minIndex = compIndex
            
    if minIndex != scanIndex:
        data[scanIndex], data[minIndex] = data[minIndex], data[scanIndex]
        print(data)
    
```

#### Insertion Sort
Brute force since its checking every element

Worst-case Speeds: O(n<sup>2</sup>)

Starts at starting point and move element to left or right based on if its greater than or less than

```python
data = [9, 5, 7, 4, 2, 8, 1, 10, 6, 3]

for scanIndex in range(1, len(data)):
    temp = data[scanIndex]
    
    minIndex = scanIndex
    
    while minIndex > 0 and temp < data[minIndex - 1]:
        data[minIndex] = data[minIndex - 1]
        minIndex -= 1
    
    data[minIndex] = temp
    print(data)
```

```python
def merge(left, right):
    # when left or right is empty that means its sorted 
    if not len(left):
        return left
    if not len(right):
        return right
    
    # variables needed to merge pieces
    result = []
    leftIndex = 0
    rightIndex = 0
    totalLen = len(right) + len(left)
    
    # Keep working until all items are merged
    while (len(result) < totalLen):
        
        # Perform the required comparison and merge
        if left[leftIndex] < right[rightIndex]:
            result.append(left[leftIndex])
            leftIndex+= 1;
        else:
            result.append(right[rightIndex])
            rightIndex+= 1
            
        #Add remaining elements of longer list to end
        if leftIndex == len(left) or rightIndex == len(right):
            result.extend(left[leftIndex:] or right[rightIndex:])
            break
        
    return result
    
      

def mergeSort(list):
    # Determines if list broken into indivual pieces
    if len(list) < 2:
        return list
    
    # Middle of list
    middle = len(list) // 2
     
    # Break the list into two pieces
    left = mergeSort(list[:middle])
    right = mergeSort(list[middle:])
    
    #Merge sorted pieces back into larger array
    print("Left side: ", left)
    print("Right side: ", right)
    merged = merge(left, right)
    print("Merged: ", merged)
    return merged
    
    
data = [9, 5, 7, 4, 2, 8, 1, 10, 6, 3]        
mergeSort(data)
```

##### Quicksort 

Avg sort time: O(n log n)
worst sort time: O(n<sup>2</sup>)

Best use cases: Sorting Arrays

Sorts data on the left and right side of a selected pivot point

Runs at worst case
* Data sorted in desired order
* Data set sorted in reverse order
* All the elements in the set are the same

```python
data = [9, 5, 7, 4, 2, 8, 1, 10, 6, 3]  

def partition(data, left, right):
    pivot = data[left]
    lIndex = left + 1
    rIndex = right
    
    while True:
        while lIndex <= rIndex and data[lIndex] <= pivot:
            lIndex += 1
        while rIndex >= lIndex and data[rIndex] >= pivot:
            rIndex -= 1
        if rIndex <= lIndex:
            break
        data[lIndex], data[rIndex] = data[rIndex], data[lIndex]
        print(data)
        
    data[left], data[rIndex] = data[rIndex], data[left]
    print(data)
    
    return rIndex

def quickSort(data, left, right):
    if right <= left:
        return
    else:
        pivot = partition(data, left, right)
        quickSort(data, left, pivot - 1)
        quickSort(data, pivot + 1, right)
        
    return data

quickSort(data, 0, len(data)-1)
```
