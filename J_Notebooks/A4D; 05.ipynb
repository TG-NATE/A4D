```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[1, 2, 3], [3, 4, 5], [5, 6, 7]])

a.dot(b)
```

np.dot(a, b) 
a is 2×3 and b is 3×3, so the inner dimensions match (3 and 3) and the result is 2×3.

The setup
        a (2×3)             b (3×3)              result (2×3)
     ┌           ┐       ┌           ┐        ┌              ┐
     │ 1  2  3   │       │ 1  2  3   │        │ 22  28  34   │
     │ 4  5  6   │   ·   │ 3  4  5   │   =    │ 49  64  79   │
     └           ┘       │ 5  6  7   │        └              ┘
                         └           ┘

Each result cell is the dot product of a row of a with a column of b.

Row 1 of a = [1, 2, 3]
Result cell	Column of b	Calculation	Value
[0,0]	[1, 3, 5]	1·1 + 2·3 + 3·5 = 1 + 6 + 15	22
[0,1]	[2, 4, 6]	1·2 + 2·4 + 3·6 = 2 + 8 + 18	28
[0,2]	[3, 5, 7]	1·3 + 2·5 + 3·7 = 3 + 10 + 21	34
Row 2 of a = [4, 5, 6]
Result cell	Column of b	Calculation	Value
[1,0]	[1, 3, 5]	4·1 + 5·3 + 6·5 = 4 + 15 + 30	49
[1,1]	[2, 4, 6]	4·2 + 5·4 + 6·6 = 8 + 20 + 36	64
[1,2]	[3, 5, 7]	4·3 + 5·5 + 6·7 = 12 + 25 + 42	79

```python
## Reshaping a matrix
    # Sometimes this is neccesary to reshape vector data as matrix so it the right shape for the algorithm
    
changeIt = np.array([1, 2, 3, 4, 5, 6, 7, 8,])

print(changeIt)

print("\n", changeIt.reshape(2, 4))

print("\n", changeIt.reshape(2, 2, 2))
```

```python
##Transpose (Changes a matix of (m x n) to (n x m))

reshaped = changeIt.reshape(2, 4)

print(reshaped, "\n")

print(np.transpose(reshaped))
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.linalg.inv(a)

print("A \n", a, "\n")
print("The inverse of A \n", b, "\n")

np.allclose(a@b, np.identity(2))

```

```python
# Print all the permutations
from itertools import permutations
a = np.array([1, 2, 3])

for p in permutations(a):
    print(p)
```

```python
#listing all 2 element combinations
from itertools import combinations
a = np.array([1, 2, 3, 4])

for combination in combinations(a, 2):
    print(combination)
```

```python
#Random 3 from a pool of 2 element comnbinations
import random
from itertools import combinations

pool = []

a = np.array([1, 2, 3, 4])

for comb in combinations(a, 2):
    pool.append(comb)
    
random.sample(pool, 3)
```
